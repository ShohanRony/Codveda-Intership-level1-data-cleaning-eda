```python
# Step 1: Load Dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
file_path = "cleaned_stock_data.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
# Summary statistics
print("Summary Statistics:\n", df.describe())
```

```python
# Mode (most frequent values)
print("\nMode:\n", df.mode().iloc[0])
```

```python
# Correlation matrix
correlation = df[['open', 'high', 'low', 'close', 'volume']].corr()
print("\nCorrelation Matrix:\n", correlation)
```

```python
# Heatmap of correlations
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Histograms
df[['open', 'high', 'low', 'close', 'volume']].hist(bins=30, figsize=(12, 8))
plt.suptitle("Histograms of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Boxplots
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[['open', 'high', 'low', 'close']])
plt.title("Boxplot of Price Features")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot: Open vs Close
plt.figure(figsize=(8, 6))
sns.scatterplot(x='open', y='close', data=df, alpha=0.5)
plt.title("Open vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")
plt.tight_layout()
plt.show()
```
