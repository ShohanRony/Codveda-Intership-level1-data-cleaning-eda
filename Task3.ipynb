```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "cleaned_stock_data.csv"
df = pd.read_csv(file_path)

# Convert 'date' back to datetime (in case it’s not)
df['date'] = pd.to_datetime(df['date'])

# Filter data for a single stock for clarity in line/bar charts
apple_df = df[df['symbol'] == 'AAPL'].sort_values('date').tail(30)
```

```python
# --- Line Chart: Closing price over time ---
plt.figure(figsize=(10, 5))
plt.plot(apple_df['date'], apple_df['close'], marker='o', linestyle='-', color='blue')
plt.title("Apple (AAPL) - Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("line_chart_close_price.png")
plt.show()
```

```python
# --- Bar Chart: Volume traded over time ---
plt.figure(figsize=(10, 5))
plt.bar(apple_df['date'], apple_df['volume'], color='green')
plt.title("Apple (AAPL) - Trading Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("bar_chart_volume.png")
plt.show()
```

```python
# --- Scatter Plot: Open vs Close Price ---
plt.figure(figsize=(8, 6))
sns.scatterplot(data=apple_df, x='open', y='close', color='purple')
plt.title("AAPL - Open vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")
plt.tight_layout()
plt.savefig("scatter_open_vs_close.png")
plt.show()
```
